- **Embedding Layer:** `token_ids → embedding_matrix[token_id]` (matrix lookup)


# 📈 SECTION 3: CALCULUS FOR AI (Derivatives & Gradient Descent)

## Why Calculus is Non-Negotiable:
Training a neural network = Finding the values of millions of parameters that minimize a "loss function." How? **Gradient Descent** — which requires computing **derivatives** (gradients).

**The core training loop of EVERY AI model (including your future GPT):**
```
1. Forward pass: prediction = model(input)          ← matrix multiplications
2. Compute loss: loss = how_wrong(prediction, truth) ← a number
3. Backward pass: gradients = d(loss)/d(weights)     ← CALCULUS (derivatives)
4. Update: weights = weights - lr * gradients         ← vector arithmetic
5. Repeat from step 1
```

You already know steps 1 and 4 from the exercises above. Now let's master step 3.

> **🔗 Salesforce Analogy:** Think of gradient descent as **Salesforce Optimizer recommendations**. The system analyzes what's slow (high loss), figures out which indexes/configs to change (gradients), and nudges them in the right direction (weight update). It repeats this until performance is optimized.

## Exercise 7: 🏔️ GRADIENT DESCENT — The Heart of AI Training

### 🎯 The Big Idea:
Imagine you're blindfolded on a mountain and need to reach the valley (lowest point). You can only feel the slope under your feet. **Gradient descent** says: "Take a step in the direction the ground slopes downward. Repeat."

- **The mountain** = your loss function (how wrong the model is)
- **Your position** = the current values of model weights
- **The slope** = the gradient (derivative)
- **Step size** = learning rate
- **The valley** = optimal weights (minimum loss)

### The Algorithm:
```
repeat until converged:
    gradient = derivative_of_loss(current_weight)
    current_weight = current_weight - learning_rate * gradient
```

That's it. That's how ChatGPT was trained. (Just with 175 billion weights instead of 1.)
